## **Q1.Create or load the dataset and display the first rows**

```python
import pandas as pd

students = [
    ["2026CS001", "Aditi", "Computer Engineering", 92, 12, 8, 84, 88],
    ["2026CS002", "Bhavana", "Computer Engineering", 64, 6, 5, 58, 61],
    ["2026IT001", "Chaitra", "Information Technology", 78, 9, 7, 72, 76],
    ["2026IT002", "Divya", "Information Technology", 55, 4, 3, 46, 52],
    ["2026CE001", "Esha", "Civil Engineering", 88, 10, 8, 79, 82],
    ["2026CE002", "Fatima", "Civil Engineering", 71, 8, 6, 66, 69],
    ["2026CS003", "Gauri", "Computer Engineering", 96, 14, 9, 91, 94],
    ["2026IT003", "Harshada", "Information Technology", 43, 2, 2, 35, 41]
]

columns = [
    "student_id", "name", "department", "attendance_pct",
    "diary_entries", "tasks_completed",
    "assignment_score", "lab_score"
]

df = pd.DataFrame(students, columns=columns)

df.to_csv("week_2_3_exam_student_signals.csv", index=False)

print(df.head())
```

### Creating the Dataset
In this step, I created the student dataset using Pandas and saved it as a CSV file. Then I displayed the first five rows to check the data.

## **Q3. Create a function that calculates final readiness score from attendance, diary, task, assignment, and lab signals.**


```python
def readiness_score(row):

    attendance = row["attendance_pct"] * 0.25

    diary = row["diary_entries"]
    if diary > 14:
        diary = 14
    diary = (diary / 14) * 20

    tasks = row["tasks_completed"]
    if tasks > 10:
        tasks = 10
    tasks = (tasks / 10) * 20

    assignment = row["assignment_score"] * 0.20
    lab = row["lab_score"] * 0.15

    total = attendance + diary + tasks + assignment + lab

    return round(total, 2)

df["readiness_score"] = df.apply(readiness_score, axis=1)

print(df[["student_id", "name", "readiness_score"]])
```

### Readiness Score
I created a function to calculate the readiness score using attendance, diary entries, completed tasks, assignment marks, and lab marks.

## **Q4. Use loops or apply logic to classify each student into Support, Developing, Strong, or Excellent.**

```python
band = []

for score in df["readiness_score"]:

    if score < 45:
        band.append("Support")

    elif score < 70:
        band.append("Developing")

    elif score < 90:
        band.append("Strong")

    else:
        band.append("Excellent")

df["band"] = band

print(df[["student_id", "name", "readiness_score", "band"]])
```

### Student Classification
Based on the readiness score, each student is placed into one of four categories: Support, Developing, Strong, or Excellent.

## **Q5. Generate at least three analytics outputs: department average, top performer, and students needing support.**

```python
print("Department Average Score")
dept_avg = df.groupby("department")["readiness_score"].mean().round(2)
print(dept_avg)

print("\nTop Performer")
top = df[df["readiness_score"] == df["readiness_score"].max()]
print(top[["student_id", "name", "readiness_score"]])

print("\nStudents Needing Support")
support = df[df["band"] == "Support"]
print(support[["student_id", "name", "department", "readiness_score"]])
```

### Analytics
I calculated the average readiness score for each department, identified the top-performing student, and listed the students who need support.
